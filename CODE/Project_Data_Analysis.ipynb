## 1. Data Descriptions and Access Methods

### Data Type 1: Species Occurrence Coordinates (GBIF API)
* **Description:** Georeferenced presence points for *Pyrus calleryana* (Callery Pear).
* **Access Method:** Programmatically retrieved via Python using the `pygbif` API library from the Global Biodiversity Information Facility (GBIF) open database (`https://api.gbif.org/`).

### Data Type 2: Bioclimatic Rasters (WorldClim / CMIP6)
* **Description:** Gridded continuous climate rasters (~1 km resolution) representing Annual Mean Temperature (Bio1), Minimum Temperature of Coldest Month (Bio6), Annual Precipitation (Bio12), and Precipitation Seasonality (Bio15).
* **Access Method:** Programmatic download from WorldClim repositories comparing present-day baseline conditions against 2050 projections (CMIP6 / SSP2-4.5).

### Data Type 3: Land Cover Data (USGS NLCD)
* **Description:** Continuous land cover raster (30-meter resolution) classifying urban, forest, agricultural, and wetland habitats.
* **Access Method:** Programmatically downloaded from the Multi-Resolution Land Characteristics (MRLC) consortium / USGS open services.

## 2. Spatial and Temporal Coverage

* **GBIF Species Occurrences:**
  * **Temporal Coverage:** 2010–2026 (filtered to match contemporary climate conditions).
  * **Spatial Coverage:** Regional Mid-Atlantic US (Bounding box: Lat 36.0°N to 39.5°N, Lon -80.5°W to -75.0°W).
* **WorldClim Climate Data:**
  * **Temporal Coverage:** Baseline vs. Mid-Century Projection (2041–2060 window).
  * **Spatial Coverage:** Continuous regional raster coverage across the Mid-Atlantic study area.
* **USGS Land Cover (NLCD):**
  * **Temporal Coverage:** 2021 NLCD release layer.
  * **Spatial Coverage:** Contiguous United States.

---

## 3. Coverage Overlap and Spatial Resolution Alignment

* **Temporal Alignment:** Occurrence records (2010–2026) closely overlap with contemporary baseline climate rasters and modern NLCD land cover classifications.
* **Resolution Harmonization Plan:**
  * WorldClim Climate Layers: ~1000-meter spatial resolution.
  * NLCD Land Cover Layer: 30-meter spatial resolution.
  * GBIF Point Data: Discrete GPS coordinates ($x, y$).
* **Data Processing Pipeline:** During feature extraction, all spatial layers will be reprojected to a common Coordinate Reference System (**WGS84 / EPSG:4326**). Raster pixel values at occurrence locations will be point-sampled using Python's `rasterio` and `geopandas` tools.

## 4. Feature Selection: Environmental Predictor Variables

To model spatial establishment risk for *Pyrus calleryana* under present-day baseline and 2050 climate scenarios, the predictive machine learning model will incorporate 6 environmental variables across climate, land cover, and topography:

1. **Bio1 (Annual Mean Temperature - °C):** WorldClim bioclimatic raster establishing the baseline thermal envelope for species survival.
2. **Bio6 (Min Temperature of Coldest Month - °C):** WorldClim bioclimatic raster capturing winter frost severity, a key physiological limit for woody plant establishment.
3. **Bio12 (Annual Precipitation - mm):** WorldClim bioclimatic raster measuring total annual moisture availability.
4. **Bio15 (Precipitation Seasonality - CV):** WorldClim bioclimatic raster quantifying rainfall variability and seasonal moisture stress.
5. **USGS NLCD Land Cover Class (Categorical):** High-resolution (30m) raster capturing human disturbance, urban corridors, agricultural edges, and natural canopy cover.
6. **Elevation / Topography (Meters):** Digital Elevation Model (DEM) data accounting for microclimate, drainage, and landscape slope.

*Note: For 2050 predictions, variables 1–4 will be replaced with CMIP6 downscaled climate projections (SSP2-4.5 scenario), while variables 5 and 6 remain static baseline predictors.*
